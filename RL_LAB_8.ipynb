```python
# Example Grid-World (replace transitions and rewards with your Assignment 7 values)

gamma = 0.9

# Each entry: state -> action -> (next_state, reward)
transitions = {
    "S0": {
        "Right": ("S1", -1),
        "Down":  ("S3", -1),
        "Up":    ("S0", -1)
    },
    "S1": {
        "Down":  ("S4", -1),
        "Left":  ("S0", -1)
    },
    "S3": {
        "Right": ("S4", -1),
        "Up":    ("S0", -1)
    },
    "S4": {
        "Up":    ("S1", -1),
        "Right": ("S4", 0)
    }
}

# Policy from the assignment
policy = {
    "S0": "Right",
    "S1": "Down",
    "S3": "Right",
    "S4": "Up"
}

def state_value_function(start_state, max_steps=100):
    """Calculate discounted return by following the policy."""
    state = start_state
    total_return = 0
    discount = 1

    for _ in range(max_steps):
        action = policy[state]
        next_state, reward = transitions[state][action]

        total_return += discount * reward
        discount *= gamma
        state = next_state

        # Stop if the path reaches a terminal state
        if state == "Terminal":
            break

    return total_return


def action_value_function(state, action, max_steps=100):
    """Take the given first action, then follow the policy."""
    next_state, reward = transitions[state][action]
    total_return = reward
    discount = gamma
    state = next_state

    for _ in range(max_steps - 1):
        if state == "Terminal":
            break

        action = policy[state]
        next_state, reward = transitions[state][action]

        total_return += discount * reward
        discount *= gamma
        state = next_state

    return total_return


print("State value V(S0):", state_value_function("S0"))
print("Action value Q(S0, Up):", action_value_function("S0", "Up"))
```

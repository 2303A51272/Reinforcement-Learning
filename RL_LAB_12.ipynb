```python
# Value Iteration in a Deterministic 2x2 Grid World

gamma = 0.9
theta = 0.001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

# Transition rules
transitions = {
    "S0": {
        "Up": "S0",
        "Down": "S1",
        "Left": "S0",
        "Right": "S0"
    },
    "S1": {
        "Up": "S0",
        "Down": "S1",
        "Left": "S1",
        "Right": "G"
    }
}

# Initialize state values
V = {"S0": 0, "S1": 0, "G": 0}

def get_reward(next_state):
    if next_state == "G":
        return 100
    return -1

# Value Iteration
iteration = 0

print("Iteration-wise State Values")
print("-" * 45)

while True:
    iteration += 1
    new_V = V.copy()

    for state in ["S0", "S1"]:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]
            reward = get_reward(next_state)

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(
        abs(new_V[state] - V[state])
        for state in ["S0", "S1"]
    )

    V = new_V

    print(
        f"Iteration {iteration}: "
        f"S0 = {V['S0']:.3f}, "
        f"S1 = {V['S1']:.3f}"
    )

    if delta < theta:
        break

# Calculate final action values and optimal policy
print("\nFinal Action Values Q*(s, a)")
print("-" * 45)

policy = {}

for state in ["S0", "S1"]:
    q_values = {}

    for action in actions:
        next_state = transitions[state][action]
        reward = get_reward(next_state)

        q = reward + gamma * V[next_state]
        q_values[action] = q

    print(f"\nState: {state}")

    for action, q in q_values.items():
        print(f"{action}: {q:.3f}")

    best_action = max(q_values, key=q_values.get)
    policy[state] = best_action

print("\nOptimal State Values")
for state in states:
    print(f"V*({state}) = {V[state]:.3f}")

print("\nOptimal Policy")
for state, action in policy.items():
    print(f"{state}: {action}")

print("\nOptimal Path: S0 -> S1 -> G")
```

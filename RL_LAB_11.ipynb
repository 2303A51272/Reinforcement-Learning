```python

# Policy Iteration in a Deterministic 2x2 Grid World

gamma = 0.9
theta = 0.001

states = ["S0", "S1", "G"]
actions = ["Up", "Down", "Left", "Right"]

# Initial policy
policy = {"S0": "Up", "S1": "Up"}

# Initialize state values
V = {"S0": 0.0, "S1": 0.0, "G": 0.0}

# Transition function
def get_next_state(state, action):
    if state == "S0":
        if action == "Right":
            return "S1"
        elif action == "Down":
            return "G"
        else:
            return "S0"

    elif state == "S1":
        if action == "Left":
            return "S0"
        elif action == "Right":
            return "G"
        else:
            return "S1"

    return "G"


# Reward function
def get_reward(next_state, state):
    if next_state == "G":
        return 100
    elif next_state == state:
        return -1
    else:
        return -1


# Policy Iteration
iteration = 0

while True:
    iteration += 1
    print(f"\nIteration {iteration}")

    # Policy Evaluation
    while True:
        delta = 0

        for state in ["S0", "S1"]:
            old_value = V[state]

            next_state = get_next_state(state, policy[state])
            reward = get_reward(next_state, state)

            V[state] = reward + gamma * V[next_state]

            delta = max(delta, abs(old_value - V[state]))

        if delta < theta:
            break

    print("State Values:", V)

    # Policy Improvement
    policy_stable = True

    for state in ["S0", "S1"]:
        old_action = policy[state]

        action_values = {}

        for action in actions:
            next_state = get_next_state(state, action)
            reward = get_reward(next_state, state)

            action_values[action] = reward + gamma * V[next_state]

        best_action = max(action_values, key=action_values.get)
        policy[state] = best_action

        print(f"\n{state} Action Values:", action_values)

        if old_action != best_action:
            policy_stable = False

    if policy_stable:
        break


# Final Results
print("\nFinal Optimal State Values:")
for state in states:
    print(state, ":", V[state])

print("\nOptimal Policy:")
for state in ["S0", "S1"]:
    print(state, ":", policy[state])

# Bellman Optimality Verification
print("\nBellman Optimality Verification:")

for state in ["S0", "S1"]:
    q_values = {}

    for action in actions:
        next_state = get_next_state(state, action)
        reward = get_reward(next_state, state)

        q_values[action] = reward + gamma * V[next_state]

    print(state)
    print("V*(s) =", V[state])
    print("Max Q*(s,a) =", max(q_values.values()))
    print("Q values =", q_values)
```
